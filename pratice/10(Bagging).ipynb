```python
X=df.drop('Outcome',axis=1)
y=df['Outcome']
# OR
# X=df.drop('Outcome',axis='columns')
# y=df.Outcome
```

```python
from sklearn.preprocessing import StandardScaler
# u can use MinMaxScaler
scale=StandardScaler()
X_scaled=scale.fit_transform(X)
X_scaled[0:3]
#  it will be in the form of numpy array
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X_scaled,y,test_size=0.2,stratify=y,random_state=42)
# read about stratify vvimp
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score

scores=cross_val_score(DecisionTreeClassifier(),X,y,cv=5)
scores
```

```python
scores.mean()
# 71% accuracy
```

```python
from sklearn.ensemble import BaggingClassifier
bag_model=BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.8,
    oob_score=True,
    random_state=42
)
bag_model.fit(X_train,y_train)
bag_model.oob_score_
```

```python
bag_model.score(X_test,y_test)
```

```python
scores=cross_val_score(bag_model,X,y,cv=5)
scores.mean()
```

```python
#  we got to know that then base model(Decisin Tree Classifier) the bagged model is best by seeing the score.mean
# bagging means row sampling with replacemnt(Resample and replace)
# in Random forest it uses bagging underneath along with data row and feature columns r divided
```
