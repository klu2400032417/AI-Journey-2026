```python
import numpy as np
import pandas as pd
import sklearn 
from sklearn.preprocessing import StandardScaler
data={
    'c1': [1, 3, 5, 7, 9],
    'c2': [7, 4, 35, 14, 56]
}
df=pd.DataFrame(data)
scale=StandardScaler()
scaled_data=scale.fit_transform(df)
res=pd.DataFrame(scaled_data,columns=df.columns)
print(res)
```

```python
from sklearn.preprocessing import MinMaxScaler
scale=MinMaxScaler()
data=scale.fit_transform(df)
res=pd.DataFrame(data,columns=df.columns)
print(res)
```
