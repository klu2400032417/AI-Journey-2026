```python

Cross-validation is a statistical technique used to evaluate a model’s performance on unseen data by repeatedly training and testing it on different subsets of the dataset, helping prevent overfitting and improve generalization.
What is Cross-Validation?
Cross-validation is a method to assess how well a predictive model will perform on independent data that was not used during training. It involves partitioning the dataset into complementary subsets, training the model on one subset (training set), and validating it on another (validation or test set). This process is repeated multiple times with different partitions, and the results are averaged to provide a robust estimate of model performance, reducing bias and variance in evaluation 


Why Use Cross-Validation?
Prevents overfitting: Ensures the model does not just memorize the training data but generalizes to new data 

Reliable performance estimation: Provides a more accurate measure of model accuracy or other metrics compared to a single train-test split 

Hyperparameter tuning: Helps select optimal model parameters without leaking information from the test set 

Common Cross-Validation Techniques
Holdout Validation: Splits the dataset into a training set and a test set (e.g., 70–30 or 80–20). Simple and fast but may be sensitive to how the split is made 

K-Fold Cross-Validation: Divides the dataset into k equal folds. The model is trained on k-1 folds and tested on the remaining fold. This is repeated k times, with each fold used once as the test set. The final performance is averaged across folds 

.
Stratified K-Fold: Ensures each fold maintains the same class distribution as the full dataset, which is especially useful for imbalanced datasets 

Leave-One-Out Cross-Validation (LOOCV): Uses all data points except one for training and tests on the single remaining point. Repeated for each data point. Provides low bias but can be computationally expensive 

Repeated K-Fold: Repeats k-fold cross-validation multiple times with different random splits to reduce variability and improve reliability 
```
