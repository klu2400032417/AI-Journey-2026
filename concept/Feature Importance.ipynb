```python
| Model               | Feature Importance     |
| ------------------- | ---------------------- |
| Linear Regression   | `coef_`                |
| Logistic Regression | `coef_`                |
| Decision Tree       | `feature_importances_` |
| Random Forest       | `feature_importances_` |
| Extra Trees         | `feature_importances_` |
| AdaBoost            | `feature_importances_` |
| Gradient Boosting   | `feature_importances_` |
| XGBoost             | `feature_importances_` |
| LightGBM            | `feature_importances_` |
| Linear SVM          | `coef_`                |
| Kernel SVM          | Not available          |
| KNN                 | Not available          |
| Naive Bayes         | No direct importance   |
| Neural Network      | No direct importance   |


EX:
lr = LinearRegression()
lr.fit(X_train, y_train)
lr.coef_

rf = RandomForestClassifier()
rf.fit(X_train, y_train)
rf.feature_importances_
```
