```python
| Model                              | Important Hyperparameters                                                                         |
| ---------------------------------- | ------------------------------------------------------------------------------------------------- |
| **Linear Regression**              | `fit_intercept`, `positive`, `n_jobs`                                                             |
| **Logistic Regression**            | `C`, `penalty`, `solver`, `max_iter`, `class_weight`, `tol`                                       |
| **KNN Classifier/Regressor**       | `n_neighbors`, `weights`, `metric`, `p`                                                           |
| **Decision Tree**                  | `max_depth`, `min_samples_split`, `min_samples_leaf`, `criterion`, `max_features`                 |
| **Random Forest**                  | `n_estimators`, `max_depth`, `min_samples_split`, `min_samples_leaf`, `max_features`, `bootstrap` |
| **SVM (SVC)**                      | `C`, `kernel`, `gamma`, `degree`, `max_iter`                                                      |
| **Naive Bayes (GaussianNB)**       | `var_smoothing`                                                                                   |
| **AdaBoost**                       | `n_estimators`, `learning_rate`, `estimator`                                                      |
| **Gradient Boosting**              | `n_estimators`, `learning_rate`, `max_depth`, `subsample`                                         |
| **XGBoost**                        | `n_estimators`, `learning_rate`, `max_depth`, `subsample`, `colsample_bytree`, `gamma`            |
| **LightGBM**                       | `n_estimators`, `learning_rate`, `num_leaves`, `max_depth`, `min_child_samples`                   |
| **CatBoost**                       | `iterations`, `learning_rate`, `depth`, `l2_leaf_reg`                                             |
| **Ridge Regression**               | `alpha`, `solver`                                                                                 |
| **Lasso Regression**               | `alpha`, `max_iter`, `tol`                                                                        |
| **ElasticNet**                     | `alpha`, `l1_ratio`, `max_iter`                                                                   |
| **PCA**                            | `n_components`, `svd_solver`                                                                      |
| **K-Means**                        | `n_clusters`, `init`, `max_iter`, `n_init`                                                        |
| **DBSCAN**                         | `eps`, `min_samples`, `metric`                                                                    |
| **Neural Network (MLPClassifier)** | `hidden_layer_sizes`, `activation`, `solver`, `learning_rate_init`, `max_iter`, `batch_size`      |






  Most Important Hyperparameter to Remember
Linear Regression → fit_intercept
Logistic Regression → C
KNN → n_neighbors
Decision Tree → max_depth
Random Forest → n_estimators, max_depth
SVM → C, kernel, gamma
XGBoost → n_estimators, learning_rate, max_depth
K-Means → n_clusters
PCA → n_components
```
